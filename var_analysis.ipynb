```python
sns.histplot(y, bins=21)
```

```python
# дисперсия
x, y 
```

```python
import numpy as np
```

```python
d_x = np.var(x)
d_x
```

```python
d_y = np.var(y)
d_y
```

```python
# среднеквадратичное отклонение
s_x = np.std(x)
s_y =np.std(y)
s_x, s_y
```

```python
np.sqrt(d_x), np.sqrt(d_y)
```

```python
# размах
np.min(x), np.max(x)
```

```python
# квартили
q1 = np.quantile(x, 0.25)
q1
```

```python
q3 = np.quantile(x, 0.75)
q3
```

```python
# межквартильный размах
q3 - q1
```

```python
import pandas as pd
```

```python
data = pd.read_csv('data.csv')
data.head()
```

```python
data.describe()
```

```python
# дисперсия
np.var(data.Sales)
```

```python
# среднеквадратичное отклонение
np.std(data.Sales)
```

```python
# размах
np.max(data.Sales) - np.min(data.Sales)
```

```python
# межквартильный размах
q1 = np.quantile(data.Sales, 0.25)
q3 = np.quantile(data.Sales, 0.75)
q3 - q1

```

```python
p1 = np.percentile(data.Sales, q=25)
p3 = np.percentile(data.Sales, q=75)
p3 - p1
```

```python
q3
```

```python
data_cut = data[data.Sales < 300]
data_cut.head()
```
