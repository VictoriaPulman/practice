```python
import numpy as np
from scipy import stats
```

```python
x = [10, 2, 2, 5, 26, 6]
```

```python
# Выборочное среднее
mean = np.mean(x)
print(f'Среднее значение массива x: {mean} ')
```

```python
# ИЛИ 
mean = sum(x) / len(x)
mean
```

```python
# выборочное среднее сильно чувствительно к выбросам
y = [10, 2, 2, 5, 26, 6, 500]
mean = np.mean(y)
print(f'Среднее значение массива y: {mean} ')
```

```python
# усеченое среднее
stats.trim_mean(y, 0.2)
print(f'Усеченое cреднее значение массива y: {stats.trim_mean(y, 0.2)} ')
```

```python
# взвешенное среднее
y = np.array([10, 2, 2, 5, 26, 6, 500])
w = np.array([0.2, 0.5, 0.5, 0.4, 0.2, 0.3, 0.1])

wm = y.dot(w) / sum(w)
wm


```

```python
# или
wm = np.average(y, weights=w)
wm
```

```python
# медиана
y = [10, 2, 2, 5, 26, 6, 500]

sorted(y)
```

```python
# медиана должна быть 6
median = np.median(y)
print(f'Медиана массива y: {median} ')
```

```python
# медиана для четного количества элементов
x = [10, 2, 2, 5, 26, 6]
median = np.median(x)
print(sorted(x))
print(f'Медиана массива x: {median} ')
```

```python
# мода 
x = [10, 2, 2, 5, 26, 6]

mode = stats.mode(x)
print(f'Мода массива x: {mode} ')
```

```python
# для бинарных значений
b = [0, 1, 0, 1, 0]
mode = stats.mode(b)
print(f'Мода массива b: {mode} ')
```
