```python
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt

import plotly.express as px

from scipy.stats import shapiro
```

```python

# Для первого файла
x_1 = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y_1 = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
pd.DataFrame({'x': x_1, 'y': y_1}).to_csv('datas/anscombe_1.csv', index=False)

# Для второго файла
x_2 = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y_2 = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
pd.DataFrame({'x': x_2, 'y': y_2}).to_csv('datas/anscombe_2.csv', index=False)

# Для третьего файла
x_3 = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y_3 = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]
pd.DataFrame({'x': x_3, 'y': y_3}).to_csv('datas/anscombe_3.csv', index=False)

# Для четвертого файла
x_4 = [8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 19.0, 8.0, 8.0, 8.0]
y_4 = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
pd.DataFrame({'x': x_4, 'y': y_4}).to_csv('datas/anscombe_4.csv', index=False)
```

```python
data_1 = pd.read_csv('datas/anscombe_1.csv')
data_2 = pd.read_csv('datas/anscombe_2.csv')
data_3 = pd.read_csv('datas/anscombe_3.csv')
data_4 = pd.read_csv('datas/anscombe_4.csv')
```

```python
data = pd.read_csv('datas/diamonds_prices.csv', index_col = 0)
data.head()
```

```python
data.info()
```

```python
data = data[:5000]
```

# Бриллианты 

```python
sns.pairplot(data);
```

# Диаграмма рассеяния

```python
sns.scatterplot(data=data, x='x', y='carat')
```

```python
sns.scatterplot(data=data, x='x', y='carat', hue='color')
```

# Гистограмма

```python
sns.histplot(data.depth, binwidth=0.5)
```

```python
data.depth.min(), data.depth.max(), data.depth.mean()
```

```python
shapiro(data.depth)
```

# Ящик с усами

```python
sns.boxplot(data=data, x = 'carat')
```

```python
sns.boxplot(data=data, x = 'carat', y = 'cut', hue='cut', legend=False)
```

# Столбчатая диаграмма

```python
data.price.mean()
```

```python
sns.barplot(data=data, x='price', y='cut', hue='cut', legend=False)
```

```python
sns.countplot(data=data, x='cut', hue='cut', legend=False)
```

# Тепловая карта

```python
number = ['price', 'depth', 'table', 'x', 'y', 'z', 'carat',]
data[number].corr()
```

```python
sns.heatmap(data[number].corr(), vmin = -1, vmax = 1, annot = True)
```

# Matplotlib

```python
plt.scatter(data.x, data.carat)

plt.xlabel('x')
plt.ylabel('carat')
```

```python
plt.scatter(data[data.cut == 'Ideal'].x, data[data.cut == 'Ideal'].carat, label = 'Ideal')
plt.scatter(data[data.cut == 'Good'].x, data[data.cut == 'Good'].carat, label = 'Good')

plt.xlabel('x', size=15)
plt.ylabel('carat', size=15)
plt.legend()
```
